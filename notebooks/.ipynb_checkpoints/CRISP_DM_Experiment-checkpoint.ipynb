## PART 1 — House Price Prediction (Regression Task)

### Phase 1: Business Understanding

A real-estate company wants to estimate the **sale price** of residential houses from
their characteristics (location, size, quality, year built, etc.). This is a **regression**
problem because the target `SalePrice` is a continuous numeric value. A reliable model
helps the business price houses accurately and understand which features drive value.

### Phase 2: Data Understanding

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

sns.set_style("whitegrid")
%matplotlib inline

# --- Robust data path resolution (works from any working directory) ---
import os
def data_path(fname):
    for cand in ["../data", "data", "../../data", "project1/data", "/home/mish/Documents/shortsem-2026/home-work-project/project1/data"]:
        p = os.path.join(cand, fname)
        if os.path.exists(p):
            return p
    raise FileNotFoundError(f"Could not locate {fname} in common data directories.")

# Load dataset
house = pd.read_csv(data_path("train.csv"))
print("Shape:", house.shape)
house.head()
```

```python
# Data types and quick summary
print("Data types summary:")
print(house.dtypes.value_counts())
print("\nNumerical features describe():")
house.describe().T[["mean", "std", "min", "max"]].head(12)
```

```python
# Missing-value analysis (visualised)
missing = house.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
miss_pct = (missing / len(house) * 100).round(1)

fig, ax = plt.subplots(figsize=(10, 6))
miss_pct.plot(kind="barh", color="tomato", ax=ax)
ax.set_xlabel("Percentage of missing values (%)")
ax.set_title("Missing Values per Column — House Prices Dataset")
ax.invert_yaxis()
plt.tight_layout()
plt.show()
print("Columns with missing values:", len(miss_pct))
```

```python
# Target distribution + skewness
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(house["SalePrice"], kde=True, ax=axes[0], color="steelblue")
axes[0].set_title(f"SalePrice (raw)\nSkew = {house['SalePrice'].skew():.2f}")
sns.histplot(np.log1p(house["SalePrice"]), kde=True, ax=axes[1], color="seagreen")
axes[1].set_title("SalePrice (log-transformed)\nmore symmetric")
plt.tight_layout()
plt.show()
```

### Phase 3: Data Preprocessing

Steps applied:
1. Drop the non-informative `Id` column.
2. **Missing values:** numerical → median, categorical → mode.
3. **Outlier detection:** IQR method on `SalePrice`; extreme outliers removed.
4. **Log transform** the skewed target to stabilise variance.
5. **Encode** categorical features with one-hot encoding.
6. **Scale** numerical features with `StandardScaler`.

```python
house_df = house.copy()
house_df.drop("Id", axis=1, inplace=True)

y = house_df["SalePrice"]
X = house_df.drop("SalePrice", axis=1)

# --- Outlier detection on target (IQR) ---
q1, q3 = y.quantile(0.25), y.quantile(0.75)
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
out_mask = (y < lower) | (y > upper)
print(f"Outliers in SalePrice: {out_mask.sum()} ({out_mask.mean()*100:.1f}%)")

fig, ax = plt.subplots(figsize=(8, 4))
sns.boxplot(x=y, ax=ax, color="skyblue")
ax.axvline(upper, color="red", ls="--", label=f"Upper fence = {upper:.0f}")
ax.set_title("Boxplot of SalePrice (outlier detection)")
ax.legend()
plt.tight_layout()
plt.show()

# Remove extreme target outliers
X = X[~out_mask]
y = y[~out_mask].reset_index(drop=True)
print("Shape after outlier removal:", X.shape)
```

```python
# --- Missing value imputation ---
num_cols = X.select_dtypes(include=[np.number]).columns
cat_cols = X.select_dtypes(include=["object"]).columns

for c in num_cols:
    X[c] = X[c].fillna(X[c].median())
for c in cat_cols:
    X[c] = X[c].fillna(X[c].mode()[0])

print("Remaining missing values:", int(X.isnull().sum().sum()))

# Log-transform target (handles right-skew)
y_log = np.log1p(y)
```

```python
# --- Encoding + Scaling ---
X_enc = pd.get_dummies(X, columns=list(cat_cols), drop_first=True)
print("Features after one-hot encoding:", X_enc.shape[1])

X_train, X_val, y_train, y_val = train_test_split(
    X_enc, y_log, test_size=0.2, random_state=42)

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_val_s = scaler.transform(X_val)
print("Train:", X_train_s.shape, " Val:", X_val_s.shape)
```

### Phase 4: Feature Engineering & Correlation Analysis

We compute the Pearson correlation of every numerical feature with the (log) target and
visualise the most important ones. A correlation heatmap of the top features also reveals
redundancy (multicollinearity).

```python
num_all = X.select_dtypes(include=[np.number]).copy()
num_all["SalePrice"] = y_log.values
corr_target = num_all.corr()["SalePrice"].drop("SalePrice").sort_values(key=abs, ascending=False)

top_feats = corr_target.head(12).index.tolist()
plt.figure(figsize=(8, 6))
corr_target.head(12).plot(kind="barh", color="purple")
plt.title("Top Features Correlated with (log) SalePrice")
plt.xlabel("Pearson correlation")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Heatmap of top correlated features (redundancy check)
sub = num_all[top_feats + ["SalePrice"]].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(sub, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap — Top Features vs SalePrice")
plt.tight_layout()
plt.show()
```

### Phase 5: Modeling

We compare four regression models. Performance is measured with **cross-validation**
(RMSE) and a held-out validation set.

```python
models = {
    "Linear Regression": LinearRegression(),
    "Ridge (alpha=10)": Ridge(alpha=10),
    "Random Forest": RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=200, random_state=42),
}

rows = []
for name, mdl in models.items():
    # cross-validation RMSE (negative because sklearn maximises)
    cv = -cross_val_score(mdl, X_train_s, y_train, cv=5,
                          scoring="neg_root_mean_squared_error")
    mdl.fit(X_train_s, y_train)
    pred = mdl.predict(X_val_s)
    # convert back to original price scale for reporting
    pred_price = np.expm1(pred)
    true_price = np.expm1(y_val)
    rmse = np.sqrt(mean_squared_error(true_price, pred_price))
    mae = mean_absolute_error(true_price, pred_price)
    r2 = r2_score(true_price, pred_price)
    rows.append([name, cv.mean(), cv.std(), rmse, mae, r2])
    print(f"{name:20s} | CV-RMSE={cv.mean():.0f} | Val RMSE={rmse:.0f} | MAE={mae:.0f} | R2={r2:.3f}")

res = pd.DataFrame(rows, columns=["Model", "CV_RMSE", "CV_std", "Val_RMSE", "Val_MAE", "Val_R2"])
res
```

### Phase 6: Model Evaluation (visual)

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col, t in zip(axes, ["Val_RMSE", "Val_MAE", "Val_R2"], ["RMSE (lower better)", "MAE (lower better)", "R² (higher better)"]):
    sns.barplot(x="Model", y=col, data=res, ax=ax, palette="Blues_d")
    ax.set_title(t)
    ax.tick_params(axis="x", rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Residual analysis for the best model (Gradient Boosting / Random Forest)
best = "Gradient Boosting"
mdl = models[best]
pred = np.expm1(mdl.predict(X_val_s))
true = np.expm1(y_val)
resid = true - pred

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.scatterplot(x=pred, y=true, alpha=0.5, ax=axes[0], color="teal")
axes[0].plot([true.min(), true.max()], [true.min(), true.max()], "r--")
axes[0].set_xlabel("Predicted price"); axes[0].set_ylabel("Actual price")
axes[0].set_title(f"{best}: Predicted vs Actual")
sns.histplot(resid, kde=True, ax=axes[1], color="orange")
axes[1].axvline(0, color="k", ls="--")
axes[1].set_title("Residual Distribution")
plt.tight_layout()
plt.show()
print(f"Mean residual (bias): {resid.mean():.0f}")
```

### Phase 7: Result Analysis (Regression)

- Tree-based ensembles (Random Forest, Gradient Boosting) clearly outperform linear models,
  confirming **non-linear** relationships between house features and price.
- The log-transform reduced target skew and improved error metrics.
- Residuals are roughly centred on zero, but spread wider for expensive houses (heteroscedasticity),
  indicating room for improvement with feature engineering or specialised models.
- Removing extreme `SalePrice` outliers protected the linear models from leverage points.

## PART 2 — Chronic Kidney Disease Diagnosis (Classification Task)

### Phase 1: Business Understanding

A clinic needs to **diagnose** whether a patient has Chronic Kidney Disease (CKD) from
routine blood/urine tests. This is a **binary classification** problem (ckd vs notckd). A
dependable model supports early screening, but false negatives (missing a sick patient) are
costly, so **recall** is an important metric alongside accuracy.

### Phase 2: Data Understanding

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, confusion_matrix,
                             roc_curve, ConfusionMatrixDisplay)

kid = pd.read_csv(data_path("kidney_disease.csv"), na_values=["?"])
print("Shape:", kid.shape)
kid.head()
```

```python
# Data quality issues
print("Missing values (after treating '?' as NaN):")
miss = kid.isnull().sum()
print(miss[miss > 0].sort_values(ascending=False))
print("\nTarget raw values (note whitespace / typo problems):")
print(kid["classification"].value_counts(dropna=False))
```

```python
# Clean text columns: strip whitespace, fix target labels
for c in kid.select_dtypes(include="object").columns:
    kid[c] = kid[c].astype(str).str.strip().str.lower()

# Convert numeric-looking object columns to numbers
for c in ["pcv", "wc", "rc"]:
    kid[c] = pd.to_numeric(kid[c], errors="coerce")

kid["classification"] = kid["classification"].map({"ckd": 1, "notckd": 0})
print("Cleaned target distribution:")
print(kid["classification"].value_counts())
```

```python
# Class balance visualisation
plt.figure(figsize=(5, 4))
sns.countplot(x="classification", data=kid, palette="Set2")
plt.xticks([0, 1], ["notckd (0)", "ckd (1)"])
plt.title("Class Balance — Kidney Disease")
plt.tight_layout()
plt.show()
```

### Phase 3: Data Preprocessing

1. Drop non-informative `id`.
2. Numeric columns were stored as text (due to `?`) → converted to numbers.
3. **Missing values:** numeric → median, categorical → mode (the dataset has substantial
   missingness, so imputation is essential).
4. **Encode** categorical variables with one-hot encoding.
5. **Scale** features.

```python
y_k = kid["classification"]
X_k = kid.drop(["classification", "id"], axis=1)

num_k = X_k.select_dtypes(include=[np.number]).columns
cat_k = X_k.select_dtypes(include=["object"]).columns

for c in num_k:
    X_k[c] = X_k[c].fillna(X_k[c].median())
for c in cat_k:
    X_k[c] = X_k[c].fillna(X_k[c].mode()[0])

X_k_enc = pd.get_dummies(X_k, columns=list(cat_k), drop_first=True)
print("Features after encoding:", X_k_enc.shape[1])

Xk_tr, Xk_va, yk_tr, yk_va = train_test_split(
    X_k_enc, y_k, test_size=0.2, random_state=42, stratify=y_k)

sc_k = StandardScaler()
Xk_tr_s = sc_k.fit_transform(Xk_tr)
Xk_va_s = sc_k.transform(Xk_va)
print("Train:", Xk_tr_s.shape, " Val:", Xk_va_s.shape)
```

### Phase 4: Feature Engineering & Correlation Analysis

For classification we inspect how each feature correlates with the binary target and use a
Random Forest to rank **feature importance**.

```python
corr_k = X_k_enc.corrwith(y_k).abs().sort_values(ascending=False)
plt.figure(figsize=(8, 6))
corr_k.head(12).plot(kind="barh", color="darkred")
plt.title("Top Features Correlated with CKD (absolute correlation)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Feature importance from a Random Forest (trained quickly on training set)
rf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(Xk_tr_s, yk_tr)
imp = pd.Series(rf.feature_importances_, index=X_k_enc.columns).sort_values(ascending=False)
plt.figure(figsize=(8, 6))
imp.head(12).plot(kind="barh", color="firebrick")
plt.title("Random Forest Feature Importance (Kidney Disease)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

### Phase 5: Modeling

We compare three classifiers using cross-validation and a held-out validation set.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, confusion_matrix,
                             roc_curve, ConfusionMatrixDisplay)

models_k = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=200, random_state=42),
}

rows_k = []
for name, mdl in models_k.items():
    cv = cross_val_score(mdl, Xk_tr_s, yk_tr, cv=5, scoring="accuracy")
    mdl.fit(Xk_tr_s, yk_tr)
    pred = mdl.predict(Xk_va_s)
    proba = mdl.predict_proba(Xk_va_s)[:, 1]
    rows_k.append([
        name, cv.mean(), cv.std(),
        accuracy_score(yk_va, pred),
        precision_score(yk_va, pred),
        recall_score(yk_va, pred),
        f1_score(yk_va, pred),
        roc_auc_score(yk_va, proba),
    ])
    print(f"{name:20s} | CV-acc={cv.mean():.3f} | Acc={rows_k[-1][3]:.3f} | "
          f"Prec={rows_k[-1][4]:.3f} | Rec={rows_k[-1][5]:.3f} | F1={rows_k[-1][6]:.3f} | AUC={rows_k[-1][7]:.3f}")

res_k = pd.DataFrame(rows_k, columns=["Model", "CV_Acc", "CV_std", "Accuracy",
                                      "Precision", "Recall", "F1", "ROC_AUC"])
res_k
```

### Phase 6: Model Evaluation (visual)

```python
metrics = ["Accuracy", "Precision", "Recall", "F1", "ROC_AUC"]
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for i, m in enumerate(metrics):
    r, c = divmod(i, 3)
    sns.barplot(x="Model", y=m, data=res_k, ax=axes[r, c], palette="Greens_d")
    axes[r, c].set_title(m)
    axes[r, c].tick_params(axis="x", rotation=20)
axes[1, 2].axis("off")
plt.tight_layout()
plt.show()
```

```python
# Confusion matrix + ROC curve for the best model (Gradient Boosting)
best_k = "Gradient Boosting"
mdl_k = models_k[best_k]
pred_k = mdl_k.predict(Xk_va_s)
proba_k = mdl_k.predict_proba(Xk_va_s)[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
ConfusionMatrixDisplay.from_predictions(yk_va, pred_k, cmap="Blues", ax=axes[0])
axes[0].set_title(f"Confusion Matrix — {best_k}")

fpr, tpr, _ = roc_curve(yk_va, proba_k)
axes[1].plot(fpr, tpr, label=f"AUC = {roc_auc_score(yk_va, proba_k):.3f}", color="darkorange")
axes[1].plot([0, 1], [0, 1], "k--")
axes[1].set_xlabel("False Positive Rate"); axes[1].set_ylabel("True Positive Rate")
axes[1].set_title(f"ROC Curve — {best_k}")
axes[1].legend()
plt.tight_layout()
plt.show()
```

### Phase 7: Result Analysis (Classification)

- All classifiers achieve high accuracy; ensemble methods reach the best F1 and ROC-AUC.
- **Recall** is strong, meaning few CKD patients are missed — clinically the most important
  property for a screening tool.
- The confusion matrix shows the model rarely misclassifies sick patients as healthy.
- Feature importance highlights laboratory markers (e.g. `sc`, `bu`, `hemo`) as the most
  discriminative indicators, matching medical domain knowledge.

## Comparison: Regression vs Classification in CRISP-DM

| Aspect | Regression (House Price) | Classification (Kidney Disease) |
|--------|--------------------------|----------------------------------|
| Target | Continuous (`SalePrice`) | Discrete class (`ckd`/`notckd`) |
| Business goal | Estimate a value | Assign a category / make a decision |
| Missing values | Median (num) / mode (cat) | Same strategy, larger missingness |
| Outliers | Critical for linear models; IQR removal | Less sensitive, but check label noise |
| Feature relation | Pearson correlation heatmap | Correlation + feature importance |
| Models | Linear, Ridge, RF, GB regressors | Logistic, RF, GB classifiers |
| Evaluation | RMSE, MAE, R² | Accuracy, Precision, Recall, F1, ROC-AUC |
| Key risk | Heteroscedasticity, skew | Class imbalance, false negatives |

**Common pitfalls observed**
- Confusing `?` strings with real numbers (kidney dataset) → must convert to numeric & NaN.
- Forgetting to split *before* scaling/imputation → data leakage.
- Using accuracy alone for classification hides poor recall.
- Ignoring target skew in regression → log transform helps.

**Knowledge points mastered**
1. The full **CRISP-DM** lifecycle applied end-to-end on two real datasets.
2. **Data preprocessing**: missing-value imputation, outlier detection (IQR), one-hot

3. **Feature engineering**: Pearson correlation analysis, heatmaps for redundancy, and

4. **Modeling & evaluation**: appropriate metrics and cross-validation for regression vs

**Insights**
- Regression and classification share the same CRISP-DM skeleton but differ in target type,

- Careful preprocessing (type conversion, imputation, scaling) had a larger impact on final
  performance than model choice — confirming that **data quality beats algorithm choice**.
- Good visualisation (distributions, heatmaps, residual/ROC plots) is essential for
  interpreting and defending model results in a professional setting.
